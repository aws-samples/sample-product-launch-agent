## Lab 1i: Creating a Multi-Agent Financial Product Launch System

### Overview

Build a **multi-agent system** where specialized agents collaborate on product launches:
- **Market Research Agent** - Analyzes markets and competitors
- **Compliance Agent** - Checks regulatory requirements
- **Strategy Agent** - Develops launch strategies
- **Marketing Agent** - Creates marketing materials
- **Orchestrator Agent** - Coordinates all agents

### Why Multi-Agent?

**Benefits:**
- Specialized expertise per domain
- Parallel task execution
- Clear separation of concerns
- Easier to maintain and extend

### Step 2: Import Multi-Agent System

```python
import os
import sys
import boto3
from boto3.session import Session

# Add current directory to path for lab_helpers imports
# This works whether you're in the labs directory or parent directory
current_dir = os.getcwd()
print(f"Current directory: {current_dir}")

# If we're in the labs directory, add it to path
if current_dir not in sys.path:
    sys.path.insert(0, current_dir)
    print(f"Added to path: {current_dir}")

# If we're NOT in labs directory, try to add labs subdirectory
if not current_dir.endswith('labs'):
    labs_path = os.path.join(current_dir, 'labs')
    if os.path.exists(labs_path) and labs_path not in sys.path:
        sys.path.insert(0, labs_path)
        print(f"Added to path: {labs_path}")

print(f"Python path: {sys.path[:3]}")

from lab_helpers.ai_multi_agent_system import (
    get_ai_orchestrator_agent,
    get_all_ai_agents
)

boto_session = Session()
region = boto_session.region_name

print(f"✅ Multi-agent system imported (Region: {region})")
```

### Step 3: Initialize the Multi-Agent System

The orchestrator coordinates 5 specialized agents.

```python
# Get the orchestrator agent
orchestrator = get_ai_orchestrator_agent()

# Get all agents for inspection
all_agents = get_all_ai_agents()

print("✅ Multi-Agent System Ready")
print("\nAvailable Agents:")
for name, agent in all_agents.items():
    print(f"  - {name}: {agent.name if hasattr(agent, 'name') else 'Agent'}")
```

### Step 4: Test Individual Agents

#### Test 1: Market Research Agent

```python
market_agent = all_agents['market']
response = market_agent("Research the current auto loan market rates and trends")
print(response)
```

#### Test 2: Marketing Agent

```python
marketing_agent = all_agents['marketing']
response = marketing_agent("Create a marketing poster for a 5.99% APR auto loan targeting millennials")
print(response)
```

### Step 5: Test Orchestrator (Full Product Launch)

The orchestrator coordinates all agents for a complete product launch.

```python
response = orchestrator(
    """Launch a new auto loan product for millennials:
    - Target rate: 5.99% APR
    - Term: 60 months
    - Target audience: 25-35 year olds
    
    Please coordinate market research, compliance check, strategy, and marketing."""
)
print(response)
```

### Step 6: Test Specific Task Delegation

```python
# Market research only
response = orchestrator("Do market research for personal loans in the US market")
print(response)
```

```python
# Marketing campaign only
response = orchestrator("Create a marketing campaign for a new credit card with 2% cashback")
print(response)
```
